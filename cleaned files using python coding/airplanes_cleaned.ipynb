```python
import pandas as pd 
path = r"C:\Users\91766\Downloads\airline_booking_dataset\2_airplanes.csv"
df = pd.read_csv(path, dtype = {"airplane_code": str})
df
```

```python
print(df.shape)
df
```

```python
df.info()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df.columns
```

```python
print(df.isnull().sum())
```

```python
for c in ["airplane_code", "model"]:
    df[c] = df[c].str.strip()
```

```python
print("Full duplicated rows: ", df.duplicated().sum())
print("Duplicate airplane codes: ", df["airplane_code"].duplicated().sum())
print("Duplicate models: ", df["model"].duplicated().sum())
df = df.drop_duplicates().drop_duplicates(subset = "airplane_code")
```

```python
df["airplane_code"] = df["airplane_code"].str.upper()
df["airplane_code"]
```

```python
print(df["airplane_code"].str.match(r"^[A-Z-9]{3,4}$").all())
```

```python
num_cols = ["range_km", "speed_kmh", "seat_capacity"]
print((df[num_cols] <= 0).sum())
```

```python
df[num_cols].describe()
```
