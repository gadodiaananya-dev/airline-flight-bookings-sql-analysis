```python
import pandas as pd

path = r"C:\Users\91766\Downloads\airline_booking_dataset\5_bookings.csv"
df = pd.read_csv(path, dtype = {"book_ref": str})
df
```

```python
print(df.shape)
```

```python
df.head()
```

```python
df.info()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df.columns
```

```python
print(df.isnull().sum())
```

```python
df["book_ref"] = df["book_ref"].str.strip().str.upper()
df["book_ref"]
```

```python
df["book_date"] = pd.to_datetime(df["book_date"], errors = "coerce")
print(df["book_date"].isnull().sum())
print(df["book_date"].min(), "->", df["book_date"].max())
```

```python
print("Full duplicate rows: ", df.duplicated().sum())
print("Duplicate book_ref: ", df["book_ref"].duplicated().sum())
df = df.drop_duplicates().drop_duplicates(subset = "book_ref")
```

```python
print(df["book_ref"].str.match(r"^[A-Z-9]{6}$").all())
```

```python
print("Zero or negative amounts: ", (df["total_amount"] <= 0).sum())
df["total_amount"].describe().round(2)
```

```python
q1, q3 = df["total_amount"].quantile([0.25, 0.75])
iqr = q3 - q1
outliers = df[(df["total_amount"] > q3 + 1.5*iqr) | (df["total_amount"] < q1 - 1.5*iqr)]
print("outliers: ", len(outliers))
```

```python
df["total_amount"] = df["total_amount"].round(2)
df["total_amount"]
```

```python
print(df.shape, df.isnull().sum().sum())
```

```python
import os
out_dir = r"C:\Users\91766\Downloads\cleaned_data"
os.makedirs(out_dir, exist_ok=True)

df.to_csv(os.path.join(out_dir, "5_bookings_clean.csv"),
          index=False, date_format="%Y-%m-%d %H:%M:%S", float_format="%.2f")
print("Saved:", len(df), "rows")
```
