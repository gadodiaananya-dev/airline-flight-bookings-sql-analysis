```python
import pandas as pd
path = r"C:\Users\91766\Downloads\airline_booking_dataset\4_flights.csv"
df = pd.read_csv(path, dtype = {"flight_no": str,  "airplane_code": str})
df
```

```python
print(df.shape)
```

```python
df.head()
```

```python
df.info()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
df.columns
```

```python
print(df.isnull().sum())
```

```python
print(df[df["actual_departure"].isnull()]["status"].value_counts())
```

```python
for c in ["flight_no", "airplane_code", "departure_airport", "arrival_airport", "status"]:
    df[c] = df[c].str.strip()

df["flight_no"] = df["flight_no"].str.upper()
df["flight_no"]
```

```python
df["airplane_code"] = df["airplane_code"].str.upper()
df["airplane_code"]
```

```python
df["departure_airport"] = df["departure_airport"].str.upper()
df["departure_airport"]
```

```python
df["arrival_airport"] = df["arrival_airport"].str.upper()
df["arrival_airport"]
```

```python
df["status"] = df["status"].str.title()
df["status"]
```

```python
date_cols = ["scheduled_departure", "scheduled_arrival", "actual_departure", "actual_arrival"]
for c in date_cols:
    df[c] = pd.to_datetime(df[c], errors = "coerce")
print(df[date_cols].isnull().sum())
```

```python
print("Full duplicate rows: ", df.duplicated().sum())
print("Duplicate flight_id: ", df["flight_id"].duplicated().sum())
print("Duplicate flight_no: ", df["flight_no"].duplicated().sum())
df = df.drop_duplicates().drop_duplicates(subset = "flight_id")
```

```python
print(df["flight_no"].str.match("^[A-Z]{2}\d{3,4}$").all())
```

```python
print(df["status"].value_counts())
```

```python
print("Arrival not after departure: ", (df["scheduled_arrival"] <= df["scheduled_departure"]).sum())
print("Same arrival and departure airport: ", (df["departure_airport"] == df["arrival_airport"]).sum())
print("Actual arrival before actual departure: ", (df["actual_arrival"] < df["actual_departure"]).sum())
```

```python
duration_hr = (df["scheduled_arrival"] - df["scheduled_departure"]).dt.total_seconds() / 3600
print(duration_hr.describe().round(2))
```

```python
delay_min = (df["actual_departure"] - df["scheduled_departure"]).dt.total_seconds() / 60
print(delay_min.describe().round(1))
```

```python
airports = pd.read_csv(r"C:\Users\91766\Downloads\cleaned_data\1_airports_clean.csv")
airplanes = pd.read_csv(r"C:\Users\91766\Downloads\cleaned_data\2_airplanes_clean.csv")

bad_dep = ~df["departure_airport"].isin(airports["airport_code"])
bad_arr = ~df["arrival_airport"].isin(airports["airport_code"])
bad_pl = ~df["airplane_code"].isin(airplanes["airplane_code"])
print(bad_dep.sum(), bad_arr.sum(), bad_pl.sum())
df = df[~(bad_dep | bad_arr | bad_pl)]
```

```python
import os
out_dir = r"C:\Users\91766\Downloads\cleaned_data"
os.makedirs(out_dir, exist_ok=True)

df.to_csv(os.path.join(out_dir, "4_flights_clean.csv"),
          index=False, date_format="%Y-%m-%d %H:%M:%S")
print("Saved:", len(df), "rows")
```
