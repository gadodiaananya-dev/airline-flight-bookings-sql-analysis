```python
import pandas as pd 
path = r"C:\Users\91766\Downloads\airline_booking_dataset\6_tickets.csv"
df = pd.read_csv(path, dtype = {"ticket_no":str, "book_ref": str, "passenger_id":str})
df
```

```python
print(df.shape)
```

```python
df.head
```

```python
df.info()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df.columns
```

```python
print(df.isnull().sum())
```

```python
for c in ["ticket_no", "book_ref", "passenger_id", "passenger_name"]:
    df[c] = df[c].str.strip()

df["book_ref"] = df["book_ref"].str.upper()
df["book_ref"]
```

```python
df["passenger_id"] = df["passenger_id"].str.upper()
df["passenger_id"]
```

```python
df["passenger_name"] = df["passenger_name"].str.replace(r"\s+", " ", regex = True) 
df["passenger_name"]
```

```python
print("Full duplicate rows: ", df.duplicated().sum())
print("Duplicate ticket_no: ", df["ticket_no"].duplicated().sum())
print("Duplicate passenger_id: ", df["passenger_id"].duplicated().sum())
df = df.drop_duplicates().drop_duplicates(subset = "ticket_no")
```

```python
print(df["ticket_no"].str.match(r"^\d{13}$").all())
print(df["passenger_id"].str.match(r"^[A-Z]{2} \d{9}$").all())
print(df["book_ref"].str.match(r"^[A-Z0-9]{6}$").all())
```

```python
print(df["outbound"].value_counts())
```

```python
print(df["passenger_name"].str.match(r"^[A-Za-z .'\-]+$").all())
print()
print(df["passenger_name"].str.len().describe().round(1))
print()
print(df[df["passenger_name"] != df["passenger_name"].str.title()]["passenger_name"].head())
```

```python
df.groupby("book_ref").size().value_counts().sort_index()
```

```python
print(df.shape, df.isnull().sum())
```

```python
print("Shape            :", df.shape)
print("Total nulls      :", df.isnull().sum().sum())
print("Duplicate rows   :", df.duplicated().sum())
print("Duplicate tickets:", df["ticket_no"].duplicated().sum())
```

```python
df = df.astype({"ticket_no": str, "book_ref": str, "passenger_id": str, "passenger_name": str, "outbound": bool})
df = df.reset_index(drop=True)
df.dtypes
```

```python
print("Tickets          :", len(df))
print("Unique bookings  :", df["book_ref"].nunique())
print("Unique passengers:", df["passenger_id"].nunique())
print()
print(df["outbound"].value_counts(normalize=True).mul(100).round(1))
df.head()
```

```python
import os

out_dir = r"C:\Users\91766\Downloads\cleaned_data"
os.makedirs(out_dir, exist_ok=True)

out_path = os.path.join(out_dir, "6_tickets_clean.csv")
df.to_csv(out_path, index=False)
print("Saved to:", out_path)
```

```python
check = pd.read_csv(out_path, dtype={"ticket_no": str, "book_ref": str, "passenger_id": str})

print("Shape matches :", check.shape == df.shape)
print("Data matches  :", check.equals(df))
print(check.dtypes)
```
